```python
import os
import sys
import pandas as pd
import numpy as np
sys.path.append(os.path.abspath(".."))
import scripts.utils as utils
%reload_ext autoreload
%autoreload 2

df_raw_age_group = utils.extract("by_age_group.xls")
df_raw_age_group

```

```python
# clean age table
df_clean_age_group = utils.clean(df_raw_age_group)
row_mapping = {
    "Not Reported /": "Not Reported",
    "No Response": "Not Reported"
}

df_clean_age_group["AGE GROUP"] = df_clean_age_group["AGE GROUP"].replace(row_mapping)
df_clean_age_group = df_clean_age_group.groupby("AGE GROUP", as_index=False).sum()
df_clean_age_group
```

```python
# build dimension table
unique_age_group = (
    df_clean_age_group[["AGE GROUP"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

unique_age_group["Age_ID"] = [f"PH_AGE_{i+1:03d}" for i in unique_age_group.index]
dim_ph_age = unique_age_group[["Age_ID", "AGE GROUP"]].rename(columns={"AGE GROUP": "Age_Group"})

dim_ph_age
```

```python
# build fact table

fact_age_prep = df_clean_age_group.merge(
    dim_ph_age, left_on="AGE GROUP", right_on="Age_Group", how="left"
)

years_cols = [col for col in df_clean_age_group.columns if col not in ["AGE GROUP", "%", "TOTAL"]]
fact_ph_age = pd.melt(
    fact_age_prep,
    id_vars=["Age_ID"],
    value_vars=years_cols,
    var_name="Year",
    value_name="Emigrant_Count"
)

fact_ph_age
```
